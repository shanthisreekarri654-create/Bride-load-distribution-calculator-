```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

L = float(input("Enter bridge length L: "))
P = float(input("Enter load P: "))
x = float(input("Enter load position x from left support: "))

if x < 0 or x > L:
    print("Invalid input: x must be between 0 and L.")
else:
    print("Inputs accepted.")
```

```python
A = np.array([
    [1, 1],
    [0, L]
], dtype=float)

B = np.array([
    P,
    P * x
], dtype=float)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
X = np.linalg.inv(A) @ B

RA = X[0]
RB = X[1]

print("Inverse matrix A^-1:")
print(np.linalg.inv(A))

print("\nSupport reactions:")
print("Reaction at A =", RA)
print("Reaction at B =", RB)
```

```python
plt.figure(figsize=(10, 4))

# Bridge
plt.plot([0, L], [0, 0], linewidth=6)

# Supports
plt.scatter([0, L], [0, 0], s=150)

# Load arrow
plt.arrow(x, 2, 0, -1.7, width=0.04,
          head_width=0.3, head_length=0.3)

# Reaction arrows
plt.arrow(0, 0, 0, 1.2, width=0.03,
          head_width=0.25, head_length=0.2)

plt.arrow(L, 0, 0, 1.2, width=0.03,
          head_width=0.25, head_length=0.2)

# Labels
plt.text(0, -0.5, f"A\nRA = {RA:.2f}", ha="center")
plt.text(L, -0.5, f"B\nRB = {RB:.2f}", ha="center")
plt.text(x, 2.2, f"Load P = {P}", ha="center")

plt.xlim(-1, L + 1)
plt.ylim(-1, 3)
plt.xlabel("Bridge position")
plt.ylabel("Reaction / load")
plt.title("Bridge Load Distribution")
plt.grid(True)

plt.show()
```
